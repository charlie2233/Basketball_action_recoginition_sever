```python
from utils import BasketballDataset, BasketballDatasetTensor, returnWeights
import numpy as np
import json
import cv2

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
from torchvision import datasets, models, transforms
from torch.utils.data import DataLoader, random_split
import time
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML

import copy
from torch.utils.data import Dataset, DataLoader, Subset
```

```python
def visualize_batches(frames, labels, mean, std, normalized=False):
    
    # Reshape Data
    # (b, c, t, h, w) => (b, t, h, w, c)
    frames = frames.permute(0, 2, 3, 4, 1).clone()
    
    # unnormalize
    if normalized:
        for i, (ch_mean, ch_std) in enumerate(zip(mean, std)):
            frames[..., i] = frames[..., i] * ch_std + ch_mean
    
    # convert to numpy array
    frames_arr = (frames * 255).numpy().astype(np.uint8)
    labels_arr = labels.numpy()
    
    labels_dict = {0 : "block", 1 : "pass", 2 : "run", 3 : "dribble", 4 : "shoot", 5 : "ball in hand", 6 : "defense", 7 : "pick" , 8 : "no_action" , 9 : "walk" , 10 : "discard"}
    # get each item in batch
    for batch_num in range(frames.shape[0]):
        frames = frames_arr[batch_num]
        label = labels_arr[batch_num]
        
        # create animation
        fig = plt.figure()
        
        vidframes = []
        # resize
        for i in range(frames.shape[0]):
            vidframes.append(cv2.resize(frames[i], dsize=(frames[i].shape[1] * 3, frames[i].shape[0] * 3), interpolation=cv2.INTER_NEAREST))
            
        im = plt.figimage(vidframes[0], resize=True)
        
        # animation function
        def animate(i):
            im.set_array(vidframes[i])
            fig.suptitle(labels_dict[label.item()], fontsize=16, color='red')
            return (im, )
        
        # animate
        anim = animation.FuncAnimation(fig, animate, frames=len(vidframes), interval=60, repeat_delay=1, repeat=True)
        display(HTML(anim.to_html5_video()))
```

```python
basketball_dataset = BasketballDatasetTensor(annotation_dict="dataset/annotation_dict.json",
                                             poseData=False)
```

```python
train_dataset_size = len(basketball_dataset)
train_dataset_indices = list(range(train_dataset_size))
np.random.shuffle(train_dataset_indices)
train_idx = train_dataset_indices[:100]
test_idx = train_dataset_indices[100:200]
```

```python
train_subset = Subset(basketball_dataset, train_idx)
test_subset = Subset(basketball_dataset, test_idx)
```

```python
train_loader = DataLoader(dataset=train_subset, shuffle=False, batch_size=10)
val_loader = DataLoader(dataset=test_subset, shuffle=False, batch_size=10)
dataloaders_dict = {'train': train_loader, 'val': val_loader}
```

```python
kinetics_mean = (0.432, 0.395, 0.376)
kinetics_std = (0.228, 0.221, 0.217)

for sample in train_loader:
    frames = sample['video']
    label = sample['class']
    visualize_batches(frames, label, kinetics_mean, kinetics_std, False)
```
